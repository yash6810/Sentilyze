### 2. Run 4-Agent Quantitative Deliberation on any Stock Ticker

```python
import yfinance as yf
from src.agent_committee import convene_trading_committee

# Choose any stock ticker (e.g. NVDA, AAPL, MSFT, TSLA, AMZN)
TICKER = 'NVDA'

print(f'🏛️ Convening 4-Agent Committee for {TICKER}...')
deliberation = convene_trading_committee(TICKER, save_resolution=False)

print('='*70)
print(f'🎯 FINAL VERDICT: {deliberation.get("final_resolution")}')
print(f'📊 CONSENSUS CONVICTION: {deliberation.get("consensus_conviction_pct"):.1f}%')
print(f'📐 FRACTIONAL KELLY SIZING: {deliberation.get("kelly_allocation_pct"):.1f}%')
print(f'🎯 TAKE-PROFIT 1 (+2.5 ATR): ${deliberation.get("tp1_target"):.2f}')
print(f'🏆 TAKE-PROFIT 2 (+4.5 ATR): ${deliberation.get("tp2_target"):.2f}')
print(f'🛡️ STOP-LOSS (-1.5 ATR):    ${deliberation.get("stop_loss_target"):.2f}')
print('='*70)

print('\n📋 SPECIALIST TESTIMONIES:')
for t in deliberation.get('agent_testimonies', []):
    print(f'● {t.get("agent_name")}: [{t.get("vote")} - {t.get("conviction_score"):.0f}%]')
    print(f'  └─ {t.get("thesis")}\n')
```
